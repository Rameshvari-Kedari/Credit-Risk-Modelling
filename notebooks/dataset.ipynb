```python

import pandas as pd
from pathlib import Path

# 1. Define file paths
file_path = Path("../data/credit_risk_prediction.arff")
csv_path = Path("../data/credit_risk.csv")

# 2. Read the ARFF file and extract column names
with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

columns = [
    line.strip().split()[1]
    for line in lines
    if line.strip().upper().startswith("@ATTRIBUTE")
]

# 3. Find where the actual data starts
data_start = next(
    i for i, line in enumerate(lines)
    if line.strip().upper() == "@DATA"
)

# 4. Load the records into Pandas
df = pd.read_csv(
    file_path,
    skiprows=data_start + 1,
    header=None,
    names=columns,
    na_values=["?"]
)

# 5. Convert numeric columns to numeric data types
numeric_columns = [
    "person_age",
    "person_income",
    "person_emp_length",
    "loan_amnt",
    "loan_int_rate",
    "loan_status",
    "loan_percent_income",
    "cb_person_cred_hist_length"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 6. Save the dataset as a CSV file
df.to_csv(csv_path, index=False)

print("Dataset successfully saved as CSV!")
print("CSV file location:", csv_path.resolve())



```
